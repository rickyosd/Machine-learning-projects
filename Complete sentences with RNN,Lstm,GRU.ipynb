```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
quotes=df["quote"]
quotes.head()
```

```python
quotes=quotes.str.lower()
```

```python
import string
translator=str.maketrans(string.punctuation,' '*len(string.punctuation))
quotes=quotes.apply(lambda x:x.translate(translator))
```

```python
quotes
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
vocab_size=10000
tokenizer=Tokenizer(num_words=vocab_size)
tokenizer.fit_on_texts(quotes)
```

```python
word_index=tokenizer.word_index
print(len(word_index))
list(word_index.items())[:10]
```

```python
sequence=tokenizer.texts_to_sequences(quotes)
print(sequence)
```

```python
X=[]
y=[]
for seq in sequence:
  for i in range(1,len(seq)):
    input_seq=seq[:i]
    output_seq=seq[i]
    X.append(input_seq)
    y.append(output_seq)
```

```python
X
```

```python
y
```

Let's illustrate with a small example how `X` and `y` are formed. Imagine a simple tokenized sentence:

`simple_sequence = [1, 2, 3, 4]` (e.g., 'the quick brown fox')

For this sequence, `X` and `y` would be populated as follows:

```python
simple_sequence = [1, 2, 3, 4] # Example: ['the', 'quick', 'brown', 'fox']

simple_X = []
simple_y = []

for i in range(1, len(simple_sequence)):
    input_seq = simple_sequence[:i]
    output_seq = simple_sequence[i]
    simple_X.append(input_seq)
    simple_y.append(output_seq)

print(f"Example X: {simple_X}")
print(f"Example y: {simple_y}")
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
max_len=10
X=pad_sequences(X,maxlen=max_len)
X
```

```python
from tensorflow.keras.utils import to_categorical
y_one_hot = to_categorical(y, num_classes=vocab_size)
```

```python
print(X[20])
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,SimpleRNN,LSTM,Dense
```

```python
embedding_dim=50
rnn_units=128
```

```python
embedding_dim=50
rnn_units=128
rnn_model = Sequential()
rnn_model.add(Embedding(input_dim=vocab_size, output_dim=embedding_dim))
rnn_model.add(SimpleRNN(units=rnn_units))
rnn_model.add(Dense(units=vocab_size,activation="softmax"))
rnn_model.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
rnn_model.summary()
```

```python
embedding_dim=50
rnn_units=128
lstm_model=Sequential()
lstm_model.add(Embedding(input_dim=vocab_size,output_dim=embedding_dim))
lstm_model.add(LSTM(units=rnn_units))
lstm_model.add(Dense(units=vocab_size,activation="softmax"))
```

```python
lstm_model.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
lstm_model.summary()
```

```python
epochs=10
batch_size=128
```

```python
history_rnn = rnn_model.fit(
    X, y_one_hot,
    epochs=epochs,
    batch_size=batch_size,
    validation_split=0.1
)
```

```python
history_lstm = lstm_model.fit(
    X, y_one_hot,
    epochs=epochs,
    batch_size=batch_size,
    validation_split=0.1
)
```

```python
word_index=tokenizer.word_index
index_to_word={}
for word,index in word_index.items():
  index_to_word[index]=word
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
def predictor(model,tokenizer,text,max_len):
  text=text.lower()
  seq=tokenizer.texts_to_sequences([text])[0]
  seq=pad_sequences([seq],maxlen=max_len,padding="pre")
  pred=model.predict(seq)
  pred=model.predict(seq,verbose=0)
  pred_index=np.argmax(pred)
  return index_to_word[pred_index]
```

```python
seed_text="country"
next_word=predictor(lstm_model,tokenizer,seed_text,max_len)
print(next_word)
```

```python
def geenrate_text(model,tokenizer,seed_text,n_words,max_len):
  text=[]
  for _ in range(n_words):
    next_word=predictor(model,tokenizer,seed_text,max_len)
    if next_word=="":
      break
    seed_text+=" "+next_word
  return seed_text
```
